```python
import math
from scipy.optimize import brentq

g=9.8
M=(20000-(300*5.1*10)/math.sqrt(101),0,2000-(300*5.1)/math.sqrt(101))
GRD=(17620-120*3.6,0,1800-0.5*g*3.6*3.6)

def f(t):
    M_new=(M[0]-(300*t*10)/math.sqrt(101),0,M[2]-(300*t)/math.sqrt(101))
    GRD_new=(GRD[0],0,GRD[2]-3*t)
    l=math.sqrt((M_new[0]-GRD_new[0])**2+(M_new[1]-GRD_new[1])**2+(M_new[2]-GRD_new[2])**2)
    r=10
    return l-r
print(M)
# print(f(4.3))
t_1=brentq(f,4,4.3,xtol=1e-8)
t_2=brentq(f,4.3,4.4,xtol=1e-8)
# print(t_1)
# print(t_2)
print(-t_1+t_2)
```
