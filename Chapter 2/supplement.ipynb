```python
from supplement import solver
import matplotlib.pyplot as plt

solverer=solver(0.001,0.2,0,2,0,0,0,1,0.17)

solverer.process()

data = solverer.resultable

t = [row[0] for row in data]
a = [row[1] for row in data]
v = [row[2] for row in data]
x = [row[3] for row in data]
um = [row[4] for row in data]
xm = [row[5] for row in data]

# ---------------------- 绘图：4个子图 ----------------------
plt.figure(figsize=(14, 10))

# 子图 1：时间 t - 加速度 a
plt.subplot(2, 3, 1)
plt.plot(t, a, 'b-', linewidth=1.5, label='a(t)')
plt.xlabel('Time t')
plt.ylabel('Acceleration a')
plt.title('t - a')
plt.grid(True)
plt.legend()

# 子图 2：时间 t - 速度 v
plt.subplot(2, 3, 2)    
plt.plot(t, v, 'r-', linewidth=1.5, label='v(t)')
plt.xlabel('Time t')
plt.ylabel('Velocity v')
plt.title('t - v')
plt.grid(True)
plt.legend()

# 子图 3：时间 t - 位移 x
plt.subplot(2, 3, 3)
plt.plot(t, x, 'g-', linewidth=1.5, label='x(t)')
plt.xlabel('Time t')
plt.ylabel('Displacement x')
plt.title('t - x')
plt.grid(True)
plt.legend()

# 子图 4：a / v / x 三条曲线同图对比
plt.subplot(2, 3, 4)
plt.plot(t, a, 'b-', linewidth=1.2, label='a')
plt.plot(t, v, 'r-', linewidth=1.2, label='v')
plt.plot(t, x, 'g-', linewidth=1.2, label='x')
plt.xlabel('Time t')
plt.title('a / v / x Comparison')
plt.grid(True)
plt.legend()

plt.subplot(2, 3, 5)
plt.plot(t, um, 'y-', linewidth=1.2, label='um(t)')
plt.xlabel('Time t')
plt.ylabel('Control Signal um')
plt.title('t - um')
plt.grid(True)
plt.legend()

plt.subplot(2, 3, 6)
plt.plot(t, xm, 'm-', linewidth=1.2, label='xm(t)')
plt.xlabel('Time t')
plt.ylabel('Target Distance xm')
plt.title('t - xm')
plt.grid(True)
plt.legend()

# 整体布局调整
plt.tight_layout()
plt.show()
```
