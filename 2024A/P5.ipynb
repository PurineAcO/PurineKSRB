```python
import P4 as sp
v_total=[]
v=1.296195
for t in range(0,101):
    v_each=[]
    print("calc",t)
    sp1=sp.dragon(t-0.001,v0=v)
    sp1.solvenb()
    sp2=sp.dragon(t+0.001,v0=v)
    sp2.solvenb()
    for i in range(224):
        v_each.append((sp2.road[i]-sp1.road[i])/0.002)
    v_max=max(v_each)
    for j in range(224):
        if v_each[j]==v_max:
            index=j
    v_total.append((index,v_max))


max_item = max(enumerate(v_total), key=lambda x: x[1][1])

max_index, max_tuple = max_item
max_vmax = max_tuple[1]

print(f"最大的v_max值为: {max_vmax}")
print(f"对应的二元组为: {max_tuple}")
print(f"在v_total中的索引为: {max_index}")
```
