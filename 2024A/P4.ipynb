```python
import math
def calculate_D_coordinate(x_O2, y_O2, x_C, y_C, theta):
    # 1. 计算半径 R：O2 到 C 的距离
    R = math.sqrt((x_C - x_O2) ** 2 + (y_C - y_O2) ** 2)
    
    # 2. 计算极轴 O2C 的方向角 alpha（O2C 与 x 轴的夹角）
    delta_x = x_C - x_O2
    delta_y = y_C - y_O2
    alpha = math.atan2(delta_y, delta_x)  # 范围: [-π, π]，atan2(y, x) 更准确
    
    # 3. 计算点 D 的总极角（极轴方向角 alpha + 已知角 theta）
    total_theta = alpha + theta
    
    # 4. 极坐标转直角坐标，得到 D 的坐标
    x_D = x_O2 + R * math.cos(total_theta)
    y_D = y_O2 + R * math.sin(total_theta)
    
    return (x_D, y_D)

# 示例：假设 O2(0,0)，C(1,0)，theta=math.pi/2（90度）
x_O2, y_O2 = sper.O2[0], sper.O2[1]
x_C, y_C =sper.C[0], sper.C[1]
theta = (13.6212-sper.CR1)/sper.r2 

x_D, y_D = calculate_D_coordinate(x_O2, y_O2, x_C, y_C, theta)
print(sper.CR1+sper.CR2)
print(f"D 的坐标为: ({x_D:.2f}, {y_D:.2f})")  # 输出: (0.00, 1.00)，符合预期
```
